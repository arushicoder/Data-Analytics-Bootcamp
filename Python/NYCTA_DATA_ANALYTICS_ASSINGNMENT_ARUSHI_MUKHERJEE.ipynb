```python
# Create a list of 10 scores
scores = [78, 85, 92, 67, 88, 74, 95, 81, 69, 90]

# Calculate total using a loop (without using sum())
total = 0

for score in scores:
    total += score

# Calculate average
average = total / len(scores)

# Display the results
print("Scores:", scores)
print("Total Score:", total)
print("Average Score:", average)
```

3.1. Convert the math score column into a NumPy array. Display its type, shape, size and first 10 values.  

```python
# Import NumPy
import numpy as np

# Convert the math score column into a NumPy array
math_scores = np.array(df["math score"])

# Display the required information
print("Type:", type(math_scores))
print("Shape:", math_scores.shape)
print("Size:", math_scores.size)
print("First 10 values:", math_scores[:10])
```

3.2. Calculate the mean, minimum, maximum, standard deviation and range of the math scores using NumPy.

```python
# Calculate statistics using NumPy

mean_score = np.mean(math_scores)
minimum_score = np.min(math_scores)
maximum_score = np.max(math_scores)
standard_deviation = np.std(math_scores)
range_score = np.max(math_scores) - np.min(math_scores)

# Display the results
print("Mean:", mean_score)
print("Minimum:", minimum_score)
print("Maximum:", maximum_score)
print("Standard Deviation:", standard_deviation)
print("Range:", range_score)
```

5.4. Grouped or side-by-side bar chart: Compare subject-wise mean scores for students who completed vs. did
not complete the test preparation course.

```python
grouped_means = df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()

subjects = ['Mathematics', 'Reading', 'Writing']

x = range(len(subjects))
width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(
    [i - width/2 for i in x],
    grouped_means.loc['none'],
    width,
    label='Did Not Complete',
    edgecolor='black'
)

plt.bar(
    [i + width/2 for i in x],
    grouped_means.loc['completed'],
    width,
    label='Completed',
    edgecolor='black'
)

plt.title('Mean Subject Scores by Test Preparation Course')
plt.xlabel('Subject')
plt.ylabel('Mean Score')
plt.xticks(x, subjects)
plt.legend()

plt.show()
```

Observation:
Students who completed the test preparation course have higher mean scores in mathematics, reading, and writing than students who did not complete the course. The difference is particularly noticeable in reading and writing scores.

6.1. Write exactly five meaningful insights from your analysis. Each insight must refer to evidence from a

```python
# Calculate the mean score for each subject
mean_math = df['math score'].mean()
mean_reading = df['reading score'].mean()
mean_writing = df['writing score'].mean()

# Calculate mean scores based on test preparation course
course_means = df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()

# Calculate the correlation between mathematics and reading
correlation = df['math score'].corr(df['reading score'])

# Calculate the number of students who completed and did not complete
course_counts = df['test preparation course'].value_counts()

# Calculate the overall average score
df['average_score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
) / 3

overall_average = df['average_score'].mean()

print("Mean Mathematics Score:", round(mean_math, 2))
print("Mean Reading Score:", round(mean_reading, 2))
print("Mean Writing Score:", round(mean_writing, 2))

print("\nMean Scores by Test Preparation Course:")
print(course_means.round(2))

print("\nCorrelation between Math and Reading:",
      round(correlation, 2))

print("\nTest Preparation Course Counts:")
print(course_counts)

print("\nOverall Average Score:",
      round(overall_average, 2))
```

Five observations

1. Subject-wise performance:
The mean reading score is higher than the mean mathematics score, while the writing score falls between them. This indicates that students performed relatively better in reading than in mathematics.

2. Effect of test preparation:
Students who completed the test preparation course have higher average mathematics, reading, and writing scores than students who did not complete the course.

3. Mathematics and reading relationship:
The correlation calculation shows a positive relationship between mathematics and reading scores. Therefore, students who perform well in mathematics generally tend to perform well in reading too.

4. Overall academic performance:
The calculated overall average score provides a summary of student performance across all three subjects. It shows that the students' combined performance is generally stronger when all three subjects are considered together.

5. Test preparation participation:
The value_counts() calculation shows the number of students who completed and did not complete the test preparation course. Comparing these groups with the grouped bar chart shows that course completion is associated with better subject-wise mean scores.

6.2. Write one recommendation for a teacher or school based on the dataset. State clearly that this dataset
alone does not establish cause and effect.  

```python
# Calculate mean scores based on test preparation course
course_means = df.groupby('test preparation course')[
    ['math score', 'reading score', 'writing score']
].mean()

print("Mean scores by test preparation course:")
print(course_means.round(2))

# Calculate the difference between completed and none
difference = (
    course_means.loc['completed'] -
    course_means.loc['none']
)

print("\nDifference in mean scores:")
print(difference.round(2))
```

Evidence:
The grouped mean scores show that students who completed the test preparation course achieved higher average scores in mathematics, reading, and writing than those who did not complete it.

Important limitation:
However, this dataset alone does not establish cause and effect. The higher scores among students who completed the course do not prove that the course itself caused the improvement, because other factors may also have influenced students' performance.
