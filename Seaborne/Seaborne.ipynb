```python
# Step 2: Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Step 3: Create sample dataset
data = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Sales': [200, 250, 300, 280, 350, 400, 420]
}

df = pd.DataFrame(data)
```

```python
# Step 4: Plot line graph
sns.set(style="whitegrid")
sns.lineplot(x='Day', y='Sales', data=df, marker='o')
```

```python
# Step 5: Add title and labels
plt.title('Weekly Sales Line Graph')
plt.xlabel('Day')
plt.ylabel('Sales')
```

```python
# Step 6: Show plot
plt.show()
```

```python
# Step 1: Install seaborn (if needed)
!pip install seaborn

# Step 2: Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Step 3: Create sample dataset
data = {
    'Hours_Studied': [1, 2, 3, 4, 5, 6, 7, 8],
    'Marks_Scored': [35, 40, 50, 55, 65, 70, 75, 85]
}

df = pd.DataFrame(data)

# Step 4: Create scatter plot
sns.set(style="whitegrid")
sns.scatterplot(x='Hours_Studied', y='Marks_Scored', data=df)

# Step 5: Add title and labels
plt.title('Hours Studied vs Marks Scored')
plt.xlabel('Hours Studied')
plt.ylabel('Marks Scored')

# Step 6: Show plot
plt.show()
```

```python
# Install seaborn (if needed)
!pip install seaborn

# Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample dataset
data = {
    'Category': ['A', 'B', 'C', 'D'],
    'Values': [10, 25, 15, 30]
}

df = pd.DataFrame(data)

# Create bar plot
sns.set(style="whitegrid")
sns.barplot(x='Category', y='Values', data=df)

# Labels and title
plt.title('Bar Plot Example')
plt.xlabel('Category')
plt.ylabel('Values')

# Show plot
plt.show()
```

```python
# Step 1: Install seaborn (if needed)
!pip install seaborn

# Step 2: Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Step 3: Create sample dataset
data = {
    'Math': [85, 90, 78, 92],
    'Science': [88, 76, 95, 89],
    'English': [90, 85, 88, 92]
}

df = pd.DataFrame(data, index=['Student1', 'Student2', 'Student3', 'Student4'])

# Step 4: Create heatmap
sns.set()
sns.heatmap(df, annot=True, cmap='coolwarm')

# Step 5: Add title
plt.title('Student Scores Heatmap')

# Step 6: Show plot
plt.show()
```
